```python
from lxml import objectify
```

```python
path="/content/drive/MyDrive/Colab Notebooks/Data/examples_of_pydata_book/datasets/mta_perf/Performance_MNR.xml"
```

```python
with open(path) as f:
  parsed=objectify.parse(f)
```

```python
root=parsed.getroot()
```

```python
data=[]
```

```python
skip_fields = ["PARENT_SEQ", "INDICATOR_SEQ",
"DESIRED_CHANGE", "DECIMAL_PLACES"]
for elt in root.INDICATOR:
  el_data={}
  for child in elt.getchildren():
    if child.tag in skip_fields:
      continue
    el_data[child.tag]=child.pyval
  data.append(el_data)
```

```python
data
```

```python
perf=pd.DataFrame(data)
```

```python
perf.head()
```
